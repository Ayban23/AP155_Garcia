```python
from math import sqrt

def f(x,nmax=50):
  for i in range(nmax):
    x = sqrt(x)
  for i in range(nmax):
    x = x**2
  return x

for xin in (5., 0.5):
  xout = f(xin); print(xin, xout)

#as nmax decreases, xout becomes more accurate/near the result of the expected analytical value

```

```python
import matplotlib.pyplot as plt
import numpy as np

i = np.linspace(1, 100, 100)
def f(x):
  return (1-np.cos(x))/x**2

x = i * 0.1
y = f(x)

plt.figure(figsize = (10,5))
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'f(x) = $\dfrac {1-cos(x)}{x^2}$')

plt.show()

for xin in (0.000000012,):
    xout = f(xin); print(xin, xout)
    #the value of xout here is not 0.5, but rather, another number. Which doesnt make sense, since the input is very small, we expect to approach the limit as x --> ∞.

```

```python
import matplotlib.pyplot as plt
import numpy as np

i = np.linspace(1, 100, 100)
def f(x):
  return 2*(np.sin(x/2)/x)**2
x = i * 0.1
y = f(x)

plt.figure(figsize = (10,5))
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'f(x) = $\dfrac{2sin^{2}(x/2)}{x^{2}}$')

plt.show()
#the graphs from the code above and in here proves that both f(x) used are the same/equivalent

for xin in (0.000000012,):
    xout = f(xin); print(xin, xout) #the value of xout now is 0.5, which is the limit of f(x) as x --> ∞
```
