```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#this is the original function f(x)
```

```python
def f(x):
    return (1-np.cos(x))/x**2
```

```python
#this is its derivative
```

```python
def fprime(x):
    return (x*np.sin(x) + 2*np.cos(x) - 2)/ x**3
```

```python
#this prints the analytical answer of f'(x) at x = 0.004
```

```python
x = 0.004
xout1 = fprime(x)
print(xout1)
```

```python
#absolute error is the central difference minus the analytical derivative at x=0.004
```

```python
def fprime_abs_error(f,x,h):
    ae = np.abs((f(x+h/2) - f(x-h/2))/h - xout1)
    return ae
```

```python
#this gives us the list of all h and absolute errors of f'(x) to be plotted in the graph
```

```python
hw = [10**(-i) for i in range(1,7)]
fprime_error_abs = [fprime_abs_error(f,x,h) for h in hw]
```

```python
#this plots the graph of the first derivative of f(x)
```

```python
plt.loglog(hw,fprime_error_abs, linestyle = "--", marker = "o")
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.grid()
plt.title("Central Difference of f'(x): Absolute Error vs. h")
plt.show()
```

```python
#This is the equivalaent function, 2sin^2(x/2) all over x^2.
```

```python
def g(x):
    return 2*(np.sin(x/2)**2)/x**2
```

```python
#This is the derivative of 2sin^2(x/2) all over x^2
```

```python
def gprime(x):
    return (x*np.sin(x) - 4*(np.sin(x/2))**2)/x**3
```

```python
#This gives the analytical answer of g'(x) at x = 0.004
```

```python
xout2 = gprime(x)
print(xout2)
```

```python
#New function, the error of g'(x):
```

```python
def gprime_abs_error(g,x,h):
    ae = np.abs((g(x+h/2) - g(x-h/2))/h - xout2)
    return ae
```

```python
#this gives us the list of all h and absolute errors of g'(x) to be plotted in the graph
```

```python
gprime_error_abs = [gprime_abs_error(g,x,h) for h in hw]
```

```python
#This graphs the first derivatives of f(x) and g(x).
```

```python
plt.loglog(hw, fprime_error_abs, linestyle = "--", marker = "o", label = "f'(x)" )
plt.loglog(hw, gprime_error_abs, linestyle = "--", marker = "o", label = "g'(x)" )
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.legend()
plt.grid()
plt.title("Central Difference of f'(x) and g'(x): Absolute Error vs. h")
plt.show()
```

```python
def f(x):
    return (1-np.cos(x))/x**2
```

```python
#this is the second derivative f(x) of the function
```

```python
def fprime2(x):
    return ((x**2)*np.cos(x)- 6*np.cos(x) - 4*x*np.sin(x) + 6)/x**4
```

```python
#This analytically answers f"(x) at x = 2.0
```

```python
y = 2.0
xout3 = fprime2(y)
print(xout3)
```

```python
#absolute error is the central difference minus the analytical derivative at x=2.0
```

```python
def fprime2_abs_error(f,y,h):
    ae = np.abs((4*f(y+h/2) + 4*f(y-h/2)- 8*f(y))/h**2 - xout3)
    return ae
```

```python
#this gives us the list of all h and absolute errors of f"(x) to be plotted on the graph
```

```python
hq = [10**(-i) for i in range(1,8)]
fprime2_error_abs = [fprime2_abs_error(f,y,h) for h in hq]
```

```python
plt.loglog(hq, fprime2_error_abs, linestyle = "--", marker = "o")
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.grid()
plt.title('Central Difference of f"(x): Absolute Error vs. h')
plt.show()
```
