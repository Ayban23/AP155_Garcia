```python
import numpy as np
```

This function creates our n by n matrix.

```python
def testcreate(n,val):
    A = np.arange(val,val+n*n).reshape(n,n)
    A = np.sqrt(A)
    return A
```

This is the function for the forward substitution

```python
def forsub(L,bs):
    n = bs.size #this is the number of rows in a column-vector yung mga x0,x1,....
    xs = np.zeros(n) 
    for i in range(n):
        xs[i] = (bs[i] - L[i,:i]@xs[:i])/L[i,i]
    return xs
```

This is the function for the backward substitution.

```python
def backsub(U,bs):
    n = bs.size
    xs = np.zeros(n)
    for i in reversed(range(n)):
        xs[i] = (bs[i]-U[i,i+1:]@xs[i+1:])/U[i,i]
    return xs
```

This is the LU decomposition function.

```python
def ludec(A):
    n = A.shape[0]
    U = np.copy(A)
    L = np.identity(n)

    for j in range(n-1):
        for i in range(j+1,n):
            coeff = U[i,j]/U[j,j]
            U[i,j:] -= coeff*U[j,j:]
            L[i,j] = coeff
    return L,U
```

This is the function that solves the system of equations using LU decomposition.

```python
def lusolve(A,bs):
    L, U = ludec(A)
    ys = forsub(L,bs)
    xs = backsub(U,ys)
    return xs
```

```python
n = 4
val = 21
A = testcreate(n,val)
print(A)
```

We solve for the inverse and determinant of A using the built-in functions in numpy.

```python
print(np.linalg.inv(A))
print(np.linalg.det(A))
```

We now apply the LU decomposition function on A.

```python
L,U = ludec(A)
print(L)
print(U)
```

The code below is of the individual determinant of the triangular matrices (the product of its diagonals). The fifth line below is the det(L)det(U) = det(A)

```python
pL = L.diagonal().prod()
pU = U.diagonal().prod()
print(pL)
print(pU)
print(pU*pL)  #This is det(L)det(U) = det(A)
print(np.linalg.det(A)) #this is the determinant using the built-in function
```

Below is solving the inverse of a through LU decomposition using 

```python
def lu_inv(A):
    L , U = ludec(A)
    n = 4
    inv = np.zeros((n,n))
    M = np.identity(4)
    for i in range(n):
        bs = M[:,i]
        xs = lusolve(A,bs)
        inv[:,i] = xs
    return inv
```

```python
print(lu_inv(A)) #this is the inverse using the LU decomp func above
print(np.linalg.inv(A)) #this is the inverse using the built-in inverse function
```
