```python
import numpy as np
```

```python
#this is the function for sum of consecutive integers
```

```python
def sumofints(nmax):
    val = 0
    for i in range (1, nmax+1):
        val += i
    return val
```

```python
#this is to try the function we created above
```

```python
nmax = 10
x = sumofints(nmax)
print(nmax, x)
```

```python
#now, lets do it in matrices. Below is a test function that makes an n by n matrix.
```

```python
def testcreate(n,val):
    A = np.arange(val,val+n*n).reshape(n,n)
    A = np.sqrt(A)
    return A
```

```python
#we define the frobenius norm of a matrix as follows
```

```python
def norm(M):
    FNorm = np.sqrt(np.sum(M**2))
    return FNorm
```

```python
#Given val = 24 and n = 4,10,20 and using our definition above, we print the norms of each matrix.
```

```python
val = 21
ns = [4,10,20]
for n in ns:
    M = testcreate(n,val)
    Fnorm = norm(M)
    print(Fnorm)
```

```python
#now we compute the norm using np.linalg.norm()
```

```python
for n in ns:
    M = testcreate(n,val)
    linalgnorm = np.linalg.norm(M)
    print(linalgnorm)
```

```python
for n in ns:
    M = testcreate(n,val)
    det = np.linalg.det(M)
    print(det)
```

```python
#now we compute for the frobenius norm using the trace formula.
```

```python
def norm_T(M):
    norm = np.sqrt(np.trace(M@(M.T.conj())))
    return norm

for n in ns:
    M = testcreate(n,val)
    Norm = norm_T(M)
    print(Norm)
```

```python
#we see that the norms we obtained from the three different functions are the same!
```
