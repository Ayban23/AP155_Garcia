```python
import matplotlib.pyplot as plt
import numpy as np

#this will be our variable
i = np.linspace(1, 100, 100)
```

```python
# this is f(x) = x**2
x = 0.1*i
y = (x)**2

#this dictates the size of our figure
plt.figure(figsize= (10,5), layout= 'constrained')

# this plots the f(x) = x**2
plt.plot(x, y)


plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r"f(x) = $x^2$")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# this is f(x) = cos(x) and f(x) = 1 - cos(x)
x = 0.1*i
y = np.cos(x)
z = 1 - np.cos(x)


#this dictates the size of our figure
plt.figure(figsize= (10,5), layout= 'constrained')

# this plots both functions
plt.plot(x, y, label = 'f(x) = cos(x)')
plt.plot(x, z, label = 'f(x) = 1 - cos(x)')


plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('f(x) = cos(x) and f(x) = 1 - cos(x)')
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


# this is f(x) = x**2
x = 0.1*i
y = (1-np.cos(x))/x**2

#this dictates the size of our figure
plt.figure(figsize= (10,5), layout= 'constrained')

# this plots the f(x) = x**2
plt.plot(x, y)


plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'f(x)= $\dfrac {1-cos(x)}{x^2}$')

plt.show()

```
