```python
import numpy as np
```

Let us first create a function creating a test matrix.

```python
def testcreate(n,val):
    A = np.arange(val,val+n*n).reshape(n,n)
    A = np.sqrt(A)
    return A
Mtest = testcreate(4,21)
print(Mtest)
```

This function solves the magnitude of a matrix.

```python
def mag(xs):
    return np.sqrt(np.sum(xs*xs))
```

This function solves for eigenfunction and eigenvector of a matrix A.

```python
def power(matrix, eps, kmax=10):
    test = np.ones(matrix.shape[0]) #test vector
    n_test = test/mag(test)   #ang magiging q^(k-1)
    
    for k in range(1,kmax):
        n_prev = n_test.copy() #need siya ipasok para ma take into account yung 
        test = matrix@ n_test  #eto na yung z^(k)
        n_test = test/mag(test) #etong n_test na to ang normalization ng z^(k) natin, hindi yung sa test vector
        print (k,n_test)
        if np.linalg.norm(n_test-n_prev) < eps:
            break
        
    lam = n_test@matrix@n_test  #as from the formula, yung lam an eigenvalue, then yung n ang (normalized) eigenvector
    return lam, n_test
```

This is the function for testing the eigenvalues.

```python
def testeigone(func, matrix):
    eps = [1e-1, 1e-2, 1e-4, 1e-6, 1e-8, 1e-10, 1e-12]
    
    for tol in eps:
        eigenval, eigenvec = func(matrix, tol)
        print(f"Power Method (eps = {tol}):")
        print(" "); print("Eigvenvalue:", eigenval); print("Eigvenvector:", eigenvec) 
        print("-" * 60)
```

We just crated a function that prints all the eigenvalues and eigenvectors of a matrix.

```python
def npeig(matrix):
    npeigvals, npeigvecs = np.linalg.eig(matrix)
    print(" "); print("NumpyEigenValue:", npeigvals); print("NumpyEigenvetor:", npeigvecs)
```

Now, we test these functions on the matrix Mtest.

```python
A = testeigone(power, Mtest)
B = npeig(Mtest)
```
